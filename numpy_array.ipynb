## NumPy array 

A NumPy array is an ndarray (the array can have multiple dimensions). The values are of the same type. The values are organized into a fixed number of axes; the dimensions. 

The array has three key properties:
.ndim: number of axes
.shape: (n, m) - think of matrix
.size: total number of elements in the array


Think of [] as one axis. Every [] adds one axis. Think of the number of axes as the number of dimensions. BUT, dimension doesn't count how many items / sublists you have, but how many levels of NESTING deep you go. I will go through this point further down in the notebook.

```python
matrix = np.array([[1,2,3],
                  [4,5,6]])

print(f"{matrix.shape}")

#I like to imagine it like a matrix. 
#This is a 2D array: matrix with 2 rows and 3 columns.
```

### 0-D array

```python
scalar_array = np.array(1)

print(f"The shape of the scalar_array: {scalar_array.shape}")
print(f"The dimension of the scalar_array: {scalar_array.ndim}")
print(f"The nr of elements in the scalar_array: {scalar_array.size}")
```

Think of no [] as 0 axes and therefore 0 dimensions. As this is just a scalar. So we cannot index it. 

### 1-D array

```python
one_dimension = np.array([1])

print(f"The shape of the 1D array: {one_dimension.shape}")
print(f"The dimension of 1D array: {one_dimension.ndim}")
print(f"The nr of elements in 1D array: {one_dimension.size}")
```

We have one [] so the dimension is 1: we have one axis which we call axis 0.
- why axis 0 and not 1? I'm not sure, but I like to think of it as index 0 for the first element in a native python list

In shape we have (1,). The first digit represents the length of the first axis: 1. There is no digit after the comma because we have no more axes (axis 1 is not there - remember, first axis is axis 0, second is axis 1). 

Let's try with different arrays:

```python
one_dimension_1 = np.array([1])
one_dimension_2 = np.array([1,2])
one_dimension_3 = np.array([1,2,3])

print(f"The shape of the first 1D array: {one_dimension_1.shape}")
print(f"The dimension of the first 1D array: {one_dimension_1.ndim}\n")
print(f"The shape of the second 1D array: {one_dimension_2.shape}")
print(f"The dimension of the second 1D array: {one_dimension_2.ndim}\n")
print(f"The shape of the third 1D array: {one_dimension_3.shape}")
print(f"The dimension of the third 1D array: {one_dimension_3.ndim}\n")

```

## An array of 1 element vs a 1x1 list are two different shapes. 

```python
#an array of one element: 
one_element = np.array([5])
print(f"An array of one element: {one_element}")
print(f"The dimension: {one_element.ndim}\n")


#1x1 array:
one_one = np.array([[5]])
print(f"An 1x1 array: {one_one}")
print(f"The dimension: {one_one.ndim}\n")
```

## 2-D array
We have [[]] --> 2 axes: axis 0 and axis 1. 

```python
two_dimensional = np.array([[5]])

print(f"The dimension of the 2D array: {two_dimensional.ndim}")
print(f"The shape of the 2D array: {two_dimensional.shape}\n")
```

### Comment on the code
- The dimension is 2 because we have two []: [[]]
- the shape describes the length of each axes:
    - (1,1): axis 0 has a length of 1 and axis 1 has a length of 1

The first array [] (outer-most bracket) has one element: [5]. This is axis 0.
The second array [5] (inner-most rbacket) has one element: 5. This is axis 1.

### Multiple sublists in 2D array

```python
new_array = np.array([[1,2],[3,4],[5,6]])

print(f"The dimension of the 2D array: {new_array.ndim}")
print(f"The shape of the 2D array: {new_array.shape}\n")
```

### Comment on the code
We have a 2D array because we only have two levels of nesting bevofre we hit a number: the outer-most [] and then the first element in the outer-most [] is another [], but with two elements which are numbers: 1 and 2. So we passed two [] before coming to the first number. 

Nesting level 1: outermost bracket. Nesting level 2: inner bracket [1,2]. 

#### Let me explain it more simply:
Axis 0 starts at the outermost bracket:
- [1,2] - first item
- [3,4] - second item
- [5,6] - third item

So axis 0 has 3 elements (3 rows). 

Now, looking inside any one of those rows, like [3,4]:
- [3,4] - two elements --> every row in this array has exactly 2 elements.
- **Every sublist in the same axis must have the same number of elements.** 

We have therefore (3,2). We can read it as a matrix: 3 rows and 2 columns. 

**Having 3 sublists side by side doesn't add a dimension, it just increases the length of axis 0. A dimension is added when we nest deeper, not when we add more items at the same level.**

**To find the dimension** quickly, think of it like this: "how many brackets do I open before I hit a raw number when picking one item (and going inward)".
Let's say I picked [5,6]. First I need to open the outermost bracket to access the three sublists. Then I need to open the bracket in [5,6] to hit a raw number. I opened two brackets: 2D.

#### Little note on indexing
You index the same way as you would a native list in python, but every index in one bracket instead of chain-indexing. The number of indices in the bracket reflects the number of dimenions. 
- array[x] - 1D
- array[x,y] - 2D
- array[x,y,z] - 3D

The shape-tuple tells us the range of valid indicise at each position. So just for like a native list of 3 elements, valid indicies are: 0-2.
